```python
exchange_rates = {
    'PKR': 0.0036,  # 1 PKR = 0.0036 USD
    'INR': 0.012,   # 1 INR = 0.012 USD
    'CNY': 0.14,    # 1 CNY = 0.14 USD
    'USD': 1.0,     # 1 USD = 1 USD
    'AED': 0.27     # 1 AED = 0.27 USD
}

# Columnas que procesar
columns_to_change = ['precio lanzamiento (Pakistan)', 'precio lanzamiento (India)',
                     'precio lanzamiento (China)', 'precio lanzamiento (USA)', 
                     'precio lanzamiento (Dubai)']

# convertir a USD
for col in columns_to_change:
    df[col] = df[col].astype(str)  # asegurar el tipo como string
    df[col + ' (USD)'] = df[col].apply(
        lambda x: exchange_rates.get(x.split()[0], None) * float(x.split()[1].replace(',', '')) 
        if isinstance(x, str) and len(x.split()) == 2 and x.split()[0] in exchange_rates else None
    )
df = df.drop(columns=columns_to_change)
```

```python
df.head()
```

```python
# Convert columns to string first, then remove units and convert to numbers
df['Peso'] = df['Peso'].astype(str).str.replace('g', '').astype(float)
df['RAM'] = df['RAM'].astype(str).str.split('/').str[0].str.extract(r'(\d+)').astype(float)
df['camara frontal'] = df['camara frontal'].astype(str).str.split('/').str[0].str.extract(r'(\d+\.?\d*)').astype(float)
df['camara trasera'] = df['camara trasera'].astype(str).str.split('+').str[0].str.extract(r'(\d+\.?\d*)').astype(float)
df['bateria'] = df['bateria'].astype(str).str.replace(',', '').str.replace('mAh', '').astype(float)
df['tamaño pantalla'] = df['tamaño pantalla'].astype(str).str.extract(r'(\d+\.?\d*)')[0].astype(float)
```
